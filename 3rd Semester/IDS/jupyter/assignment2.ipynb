# Problem 1 – Delivery Path Analyzer
i used list of distionaries to store multiples points with their location time and distances from previous point , total distance would be suum of distances b\w all points

```python

# Problem 1 – Delivery Path Analyzer
n = int(input("Enter number of delivery points: "))
deliveries = [] #using list of dictionaries

for i in range(n):
    loc = input(f"Enter location {i+1}: ")
    time = input(f"Enter time at {loc}: ")
    dist = float(input(f"Enter distance at {loc}: "))
    deliveries.append({"location": loc, "time": time, "distance": dist})

start = deliveries[0]["location"] #starting point
end = deliveries[-1]["location"] #ending point/destination
total_distance = sum([d["distance"] for d in deliveries]) #add all distances b/w points

print("\nStarting Location:", start)
print("Ending Location:", end)
print("Total Distance Covered:", total_distance)

```

# Problem 2 – Shopping Cart Management
it displayes a menu to user to press 1 to add item 2. to remove item 3. to update item price and 0 to end , a while loop is used until user enters 0
for 1 item name and price are taken and added in list , for 2 item to be removed name is take and removed from list, for update take name and new price at last, print totalprice of items and maximum price of an item


```python
# Problem 2 – Shopping Cart Management
def CMS():
    cart=[]
    n=1
    while(n!=0):
        n=int(input("Enter 1 to Add, 2 to Remove, 3 to update or 0 to end: "))
        if (n==1):
            item=input("Enter item name: ")
            price=float(input(f"Enter price of {item}:"))
            cart.append((item, price))
        elif (n==2):
            if not cart:
                print("cart is empty")
            else:
                item_to_remove=input("Enter name of the item to remove: ")
                for i in cart:
                    if i[0]==item_to_remove:
                        cart.remove(i)
        elif (n==3):
            if not cart:
                print("cart is empty")
            else:
                item_to_update=input("Enter name of the item to update the price: ")
                new_price=float(input("Enter new price: "))
                for i in range(len(cart)):
                    if cart[i][0] == item_to_update:  
                        cart[i] = (item_to_update, new_price) 
                    
    if not cart:
        print("cart is empty")
    else:
        total_price = sum([d[1] for d in cart])
        maximum_price=max(d[1] for d in cart)
        print(f"Maximum Price: {maximum_price}")
        print(f"Total price: {total_price}")

CMS()
```

# Problem 3 – Student Performance Report
it takes no of students you want to add then using a nested loop take student with multiple subjects then find averages of students and store it in a list along with their names making a tuple then sort this list to fin top 3 performers

```python
# Problem 3 – Student Performance Report 
def studentPerformanceReport():
    students = []
    n = int(input("Enter number of students: "))

    for i in range(n):
        name = input(f"\nEnter name of student {i+1}: ")
        num_subjects = int(input(f"Enter number of subjects for {name}: "))
        marks = []

        for j in range(num_subjects):
            score = float(input(f"Enter marks for subject {j+1}: "))
            marks.append(score)

        students.append((name, marks))
        
    averages = []
    for name, marks in students:
        avg = sum(marks) / len(marks)
        averages.append((name, avg))
        
    averages.sort(key=lambda x: x[1], reverse=True) #true = decending order

    for name, avg in averages:
        print(f"{name} - Average Marks: {avg:.2f}")
        
    print("Top 3 Performers:")
    for i, (name, avg) in enumerate(averages[:3], start=1):
        print(f"{i}. {name} - {avg:.2f}")
        
studentPerformanceReport()
```

# Problem 4 – Email Data Processor
for this problem asked user to enter number of emails, then take each email and check if it contains “@” and “.” to validate it. valid emails are stored in a list. after taking all inputs, program extracts domain names from each email, counts how many emails belong to each domain using a dictionary, and finally displays each domain with its count.

```python
# Problem 4 – Email Data Processor
def email_system():
    n=int(input("Enter number of emails: "))
    emails=[]
    for i in range(n):
        email=input(f"Enter Email {i+1}: ")
        if "@" in email and "." in email:
            emails.append(email)
        else:
            print("email is invalid.")
    domain_counts={}
    for email in emails:
        domain = email.split("@")[-1].lower()
        if domain in domain_counts:
            domain_counts[domain]+=1
        else:
            domain_counts[domain]=1
    for domain, count in domain_counts.items():
        print(f"{domain}: {count}")
email_system()
```

# Problem 5 – Friendship Network Analysis
for this problem asked user to enter names of two users and their friends as comma-separated lists. program then converts them into sets and finds mutual friends, unique friends for each user, and possible friend suggestions. finally displays mutual, unique, and suggested friends for both users

```python
# Problem 5 – Friendship Network Analysis
def friendship_network():
    user1 = input("Enter name of 1st User: ")
    user2 = input("Enter name of 2nd User: ")
    friends1 = set(input(f"Enter friends of {user1} (comma separated): ").split(","))
    friends2 = set(input(f"Enter friends of {user2} (comma separated): ").split(","))
    friends1 = {f.strip() for f in friends1}
    friends2 = {f.strip() for f in friends2}

    mutual_friends=friends1 & friends2
    unique_friends1 = friends1 - friends2
    unique_friends2 = friends2 - friends1
    if not mutual_friends:
        print("No Mutual Friends.")
    else:
        print(f"Mutual friends: {mutual_friends}")
    if not unique_friends1:
        print(f"No unique friends of {user1}.")
    else:
        print(f"Unique friends of {user1}: {unique_friends1}")
    if not unique_friends1:
        print(f"No unique friends of {user2}.")
    else:
        print(f"Unique friends of {user2}: {unique_friends2}")
    if not unique_friends2:
        print(f"No friend suggestions for {user1}.")
    else:
        print(f"Friend suggestions for {user1}: {unique_friends2}")  
    if not unique_friends1:
        print(f"No friend suggestions for {user2}.")
    else:
        print(f"Friend suggestions for {user2}: {unique_friends1}") 

friendship_network()    

```

# Problem 7 – Transport Schedule Tracker
for this problem asked user to enter number of stops, then for each stop take name, distance, and time taken and store them in a list. after taking all inputs calculate total travel time, find the minimum time to identify the fastest stop(s), then display total number of stops, total time taken, and the fastest route

```python
# Problem 7 – Transport Schedule Tracker
def transport_schedule():
    n=int(input("Enter number of stops: "))
    route=[]
    if n==0:
        return
    for i in range(n):
        print(f"Enter details of stop {i+1}: ")
        name=input("Name: ")
        distance=float(input("Distance(km): "))
        time=float(input("time_taken(min): "))
        route.append((name,distance,time))
    total_time=sum([t[2] for t in route])
    fastest=min([t[2] for t in route])
    fastest_stop=[]
    for k in route:
        if k[2]==fastest:
            fastest_stop.append(k[0])
    print(f"Total number of stops: {len(route)}")
    print(f"Total time taken: {total_time}")
    print(f"Fastest Route: {fastest_stop}")
        
    
transport_schedule()
```

# Problem 9 – Sports Tournament Statistics
for this problem asked user to enter number of matches, then for each match take names of two teams and their scores and store them in a list. after taking all inputs calculate total scores for each team using a dictionary. then find total number of teams, average score, and highest scoring teams. finally display each team with total score, overall average, and the winning teams

```python
# Problem 9 – Sports Tournament Statistics
def sports_tournament_stats():
    n = int(input("Enter number of matches: "))
    matches = []
    if n == 0:
        print("No matches.")
        return
    for i in range(n):
        print(f"Enter details of match {i+1}: ")
        team1 = input("Team 1 name: ")
        team2 = input("Team 2 name: ")
        score1 = int(input(f"Score of {team1}: "))
        score2 = int(input(f"Score of {team2}: "))
        matches.append((team1, team2, score1, score2))

    team_scores = {}
    for match in matches:
        team1 = match[0]
        team2 = match[1]
        score1 = match[2]
        score2 = match[3]

        if team1 in team_scores:
            team_scores[team1] += score1
        else:
            team_scores[team1] = score1

        if team2 in team_scores:
            team_scores[team2] += score2
        else:
            team_scores[team2] = score2
    total_score = 0
    total_teams = len(team_scores)
    for team, score in team_scores.items():
        total_score += score
    average_score = total_score / total_teams
    maximum_score =max(score for team, score in team_scores.items())
    
    winners = []
    for team, score in team_scores.items():
        if (score == maximum_score):
            winners.append(team)
    print(f"Total teams: {total_teams}")
    print(f"Average score: {average_score}")
    print("Teams and their total scores:")
    for team, score in team_scores.items():
        print(f"{team}: {score}")
    print("Winning Team(s): ", winners)

sports_tournament_stats()
```

# Problem 10 – Weather Data Analyzer
for this problem program takes a list of temperatures and finds maximum temperature and the days it occurred, then calculates average temperature of all days. it also counts how many times each temperature appeared to find the most common (usual) temperature. after that it displays average, hottest, and most common temperatures, and finally gives a message if the usual weather is cold, pleasant, or hot based on temperature range.

```python
# Problem 10 – Weather Data Analyzer
def weather_analyzer(temp):
    maximum = max(temp)
    hottest_day = []
    summ = 0
    for i in range(len(temp)):
        summ = summ + temp[i]
        if temp[i] == maximum:
            hottest_day.append(i + 1) 

    average = summ / len(temp)
    temps_counts = {}
    for t in temp:
        if t in temps_counts:
            temps_counts[t] += 1
        else:
            temps_counts[t] = 1
    max_count = 0
    usual_temp = 0
    for temp, count in temps_counts.items():
        if count > max_count:
            max_count = count
            usual_temp = temp

    print(f"Average temperature: {average}")
    print(f"Hottest temperature: {maximum} on day(s): {hottest_day}")
    print(f"Most common temperature: {usual_temp} (appeared {max_count} times)")

    if 15.0 < usual_temp < 35.0:
        print(f"Weather is usually pleasant as temp is {usual_temp}")
    elif usual_temp <= 15.0:
        print(f"Weather is usually cold as temp is {usual_temp}")
    elif usual_temp >= 35.0:
        print(f"Weather is usually hot as temp is {usual_temp}")
        
temps = [41, 31, 44, 32, 33, 31, 30, 34, 35, 33,32, 35, 31, 29, 28, 30, 32, 33, 34, 35,31, 30, 30, 33, 32, 31, 29, 34, 33, 30]
weather_analyzer(temps)
```

# Problem 11 – Employee Attendance System 
for this problem asked user to enter number of employees and number of days, then for each employee take name and attendance record for given days as P or A separated by spaces. store number of present days for each employee in a dictionary. after taking all inputs calculate attendance percentage for each employee and display it. finally find and display employee with most presence and most absences.

```python
# Problem 11 – Employee Attendance System 
def Attendance_system():
    n = int(input("Enter number of employees: "))
    if n <= 0:
        print("No employees.")
        return

    days = int(input("Enter number of days: "))
    if days <= 0:
        print("Invalid days.")
        return

    attendance = {}

    for i in range(n):
        name = input(f"Enter name of employee {i+1}: ")
        data = input(f"Enter attendance for {name} ({days} days as P or A, space separated): ").split()
        if len(data) != days:
            print("Invalid")
            return
        present = data.count("P") + data.count("p")
        attendance[name] = present

    sorted_attend = sorted(attendance.items(), key=lambda x: x[1], reverse=True)
    most_present = sorted_attend[0]
    least_present = sorted_attend[-1]

    for name, count in attendance.items():
        percent=(count/days)*100
        print(f"{name}: {percent}%")
    print(f"Perfect Attendence: {most_present[0]} ({most_present[1]})")
    print(f"Most Absences: {least_present[0]} ({least_present[1]})")

Attendance_system()
```

# Problem 12 – Fitness Tracker
for this problem asked user to enter number of users then for each user take name and number of workout sessions, for each session take session name, time and calories burnt then store them in a list (list of lists). after taking all inputs calculate total calories and average time for each user and display them. also find and display all unique activity types at the end

```python
# Problem 12 – Fitness System
def fitness_system():
    n = int(input("Enter number of users: "))
    users = []
    if n <= 0:
        print("No users.")
        return
    for i in range(n):
        name = input(f"Enter name of user {i+1}: ")
        sessions = []
        s = int(input(f"Enter number of sessions for {name}: "))
        for j in range(s):
            s_name = input(f"Session {j+1} name: ")
            s_time = float(input(f"Session {j+1} time (minutes): "))
            s_calories = float(input(f"Calories burnt in session {j+1}: "))
            sessions.append([s_name, s_time, s_calories])
        users.append([name, sessions])
    all_activities = set()

    for user in users:
        name = user[0]
        sessions = user[1]
        total_calories = 0
        total_time = 0
        for session in sessions:
            totalcalories += session[2]
            totaltime += session[1]
            all_activities.add(session[0])
        avg_time = totaltime / len(sessions)
        print(f"{name}: Total Calories Burned = {totalcalories}, Average Workout Time = {avg_time} minutes")
    print(f"Unique activity types: {all_activities}")

fitness_system()
```

# Problem 13 – Online Quiz Evaluator 
for this problem asked user to enter no of students he want to enter then take that number of inputs with name , score,time taken then store them in a list making them a list (list of lists) then taking minimum threshold from user and finding maximum of scores , after that using a loop fin in whole list for top scorers and students passed and store them in different lists at the end display average , passing and top scorer students

```python
# Problem 13 – Online Quiz System
def Quiz_System():
    n=int(input("Enter number of students: "))
    quiz=[]
    if(n<=0):
        print("Invalid.")
        return
    for i in range(n):
        name=input(f"Enter name of student {i+1}: ")
        scores=float(input(f"Enter scores of {name}: "))
        time=float(input(f"Enter time taken by {name}: "))
        quiz.append([name, scores, time])
    threshold=float(input("Enter minimum threshold: "))
    top_scorers=[]
    maxi=max(t[1] for t in quiz)
    avg=sum(i[1] for i in quiz)
    avg=avg/n
    passed=[]
    for i in quiz:
        if i[1]==maxi:
            top_scorers.append(i[0])
        if i[1]>=threshold:
            passed.append(i[0])

    print(f"Average Scores: {avg}")
    print(f"Students who passed:\n {passed}")
    print(f"Top Scorers:\n {top_scorers}")

Quiz_System()
```

# Problem 14 – Hotel Booking Management
for this problem program manages 30 hotel rooms using a dictionary where room numbers are keys and guest names are values. user can choose to add a booking (assign next available room), checkout a guest (free the room), or exit to see summary. at the end program displays list of empty rooms and occupied rooms with guest names.

```python
# Problem 14 – Hotel Booking Management
def hotel_booking_system():
    rooms = {i: None for i in range(1, 31)}
    print(f"We have 30 rooms in total.")
    choice = 1
    while choice != 0:
        choice = int(input("Enter 1 to add (give room), 2 to checkout, 0 to summary/exit: "))
        if choice == 1:
            free_room = None
            for r in range(1, 31):
                if rooms[r] is None:
                    free_room = r
                    break
            if free_room is None:
                print("No rooms available")
            else:
                guest = input("Enter guest name: ")
                rooms[free_room] = guest
                print(f"Assigned room {free_room} to {guest}")
        elif choice == 2:
            room_str = input("Enter room number to checkout (1-30): ")
            if not room_str.isdigit():
                print("Invalid room number")
            else:
                rnum = int(room_str)
                if rnum < 1 or rnum > 30:
                    print("Room number must be 1-30")
                else:
                    if rooms[rnum] is None:
                        print(f"Room {rnum} is already empty")
                    else:
                        print(f"Checked out {rooms[rnum]} from room {rnum}")
                        rooms[rnum] = None
        elif choice == 0:
            break
    empty = [str(r) for r in range(1, 31) if rooms[r] is None]
    occupied = [(r, rooms[r]) for r in range(1, 31) if rooms[r] is not None]

    if empty:
        print(f"{', '.join(empty)} rooms are empty.")
    else:
        print("No empty rooms.")
    if occupied:
        print("Occupied rooms:")
        for r, g in occupied:
            print(f" Room {r}: {g}")
    else:
        print("No occupied rooms")

hotel_booking_system()
```

# Problem 15 – Expense & Budget Tracker 
for this problem asked user to enter total budget and number of expenses, then for each expense take category and amount and store totals per category in a dictionary. after taking all inputs calculate total spent and find category with highest spending. finally display total spent, spending per category, whether user stayed within or exceeded budget, and the category with most expenses.

```python
# Problem 15 – Expense & Budget Tracker 
def financial_tracker():
    budget = float(input("Enter total budget: "))
    n = int(input("Enter number of expenses: "))
    if n <= 0:
        print("No expenses.")
        return

    category_totals = {}
    total_spent = 0.0

    for i in range(n):
        catgry = input(f"Enter category for expense {i+1}: ")
        amount = float(input(f"Enter amount for {catgry}: "))
        total_spent += amount

        if catgry in category_totals:
            category_totals[catgry] += amount
        else:
            category_totals[catgry] = amount

    max_cat = None
    max_amt = 0
    for c, a in category_totals.items():
        if a > max_amt:
            max_amt = a
            max_cat = c
    print(f"Total spent: {total_spent}")
    print("Spent per category:")
    for cat, tot in category_totals.items():
        print(f" {cat}: {tot}")

    if total_spent > budget:
        print(f"You exceeded the budget by {total_spent - budget}")
    else:
        print(f"You are within budget. Remaining: {budget - total_spent}")

    print(f"Most money spent on: {max_cat} ({max_amt})")

financial_tracker()
```

# Problem 16 – Smart Business Manager
for this problem program works as a mini business management system. it asks user to add transactions with date, type (income or expense), amount, and category and stores them in a list of dictionaries. it has separate functions to add new transactions, show total amount by category, find highest expense, and generate a full financial summary showing total income, total expenses, and net balance

```python
# Problem 16 – Smart Business Manager
def add_transaction(transactions):
    date = input("Enter date (dd-mm-yyyy): ")
    t_type = input("Enter transaction type (Income/Expense): ")
    amount = float(input("Enter amount: "))
    category = input("Enter category: ")
    transactions.append({"date": date, "type": t_type, "amount": amount, "category": category})


def total_summary(transactions):
    summary = {}
    for t in transactions:
        cat = t["category"]
        if cat in summary:
            summary[cat] += t["amount"]
        else:
            summary[cat] = t["amount"]
    print("Category-wise totals:")
    for c, a in summary.items():
        print(f" {c}: {a}")
    return summary


def highest_expense(transactions):
    max_amt = 0
    max_cat = None
    for t in transactions:
        if t["type"].lower() == "expense" and t["amount"] > max_amt:
            max_amt = t["amount"]
            max_cat = t["category"]
    if max_cat:
        print(f"Highest expense in category: {max_cat} ({max_amt})")
    else:
        print("No expenses recorded.")


def financial_report(transactions):
    income = sum(t["amount"] for t in transactions if t["type"].lower() == "income")
    expense = sum(t["amount"] for t in transactions if t["type"].lower() == "expense")
    balance = income - expense
    print("\n Financial Summary: ")
    print(f"Total Income: {income}")
    print(f"Total Expense: {expense}")
    print(f"Net Balance: {balance}")


def smart_business_manager():
    transactions = []
    while True:
        print("\n1. Add Transaction\n2. Show Summary\n3. Show Highest Expense\n4. Generate Report\n0. Exit")
        choice = int(input("Enter choice: "))
        if choice == 1:
            add_transaction(transactions)
        elif choice == 2:
            total_summary(transactions)
        elif choice == 3:
            highest_expense(transactions)
        elif choice == 4:
            financial_report(transactions)
        elif choice == 0:
            break
        else:
            print("Invalid choice.")

smart_business_manager()
```
