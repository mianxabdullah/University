```python
traffic_logs = [
"id:501,zone:A1,vehicle:Car,speed:62,time:08:30,violations:[None],status:Smooth",
"id:502,zone:A1,vehicle:Bike,speed:85,time:09:10,violations:[Helmet],status:Busy",
"id:503,zone:B2,vehicle:Bus,speed:45,time:17:25,violations:[None],status:Smooth",
"id:504,zone:C3,vehicle:Car,speed:110,time:14:15,violations:[Overspeed],status:Congested",
"id:505,zone:A1,vehicle:Truck,speed:40,time:18:50,violations:[None],status:Smooth"
]
print("Task 1: ")
def make_dict(traffic):
    traf=[]
    for t in traffic:
        try:
            parts=t.split(',')
            trf={}
            for p in parts:
                key,val=p.split(':',1)
                if key in ['id','speed']:
                    val=int(val)
                elif key == 'violations':
                    val = val.strip('[]').split(',')
                    if val == ['None']:
                        val = []
                trf[key]=val
            traf.append(trf)
        except Exception as e:
            print("Error in record:", t, "->", e)
    return traf
structured_data=make_dict(traffic_logs)
print("Data after processing into structured for: \n",structured_data)
```

```python
print("Task 2: ")
def avgSpeedPerZone(traffic):
    zone_speeds = {}  
    for t in traffic:
        zone = t['zone']
        if zone not in zone_speeds:
            zone_speeds[zone] = []
        zone_speeds[zone].append(t['speed'])
    avg_per_zone = {}
    for zone, speeds in zone_speeds.items():
        if speeds:
            avg_per_zone[zone] = sum(speeds)/len(speeds)
        else:
            avg_per_zone[zone] = None
    print("Zone Wise Avg Speed:")
    for zone,avg_speed in avg_per_zone.items():
        if avg_speed is not None:
            print(f"{zone}: {avg_speed}")
        else:
            print(f"{zone}: No data")
    return avg_per_zone
avgSpeedPerZone(structured_data)
```

```python
print("Task 3: ")
def peakHour(traffic):
    hour_counts = {}
    for t in traffic:
        hour = t['time'].split(':')[0] 
        if hour not in hour_counts:
            hour_counts[hour] = 0
        hour_counts[hour] += 1
    maxi = max(hour_counts.values())
    peak_hrs=[]
    for h,c in hour_counts.items():
        if c==maxi:
            peak_hrs.append(h)
    print("Peak hours:", peak_hrs)
    print("Number of entries:", maxi)
peakHour(structured_data)
```

```python
print("Task 4: ")
def overspeed(traffic):
    idnames={}
    for t in traffic:
        if t['speed']>80:
            idnames[t['id']]=t['vehicle']
    return idnames
print(overspeed(structured_data))
```

```python
print("Task 5: ")
def countofviolation(traffic):
    vcount={}
    for t in traffic:
        for v in t['violations']:     
            if v not in vcount:
                vcount[v]=0
            vcount[v]+= 1
    return vcount
print(countofviolation(structured_data))
```

```python
print("Task 6: ")
def safety_index(traffic):
    zones={}
    for t in traffic:
        z=t['zone']
        if z not in zones:
            zones[z]={"risk":0}
        if t['speed']>80:
            zones[z]["risk"]+=2
        violations=t['violations']
        zones[z]["risk"]+=len(violations)
        status = t['status']
        if status=="Busy":
            zones[z]["risk"]+= 1
        elif status=="Congested":
            zones[z]["risk"] +=2
    print("Safety Index per Zone:")
    for z, d in zones.items():
        ttlRisk = d["risk"]
        sindex = 100-(ttlRisk*5)
        if sindex < 0:
            sindex=0   
        print(f"{z}: {sindex}")
safety_index(structured_data)
```

```python
print("Task 7: ")
def vehicle_summary(traffic):
    summary = {}
    for t in traffic:
        vehicle = t['vehicle']
        if vehicle not in summary:
            summary[vehicle] = {"count": 0,"total_speed": 0,"violations_count": 0}
        summary[vehicle]["count"]+=1
        summary[vehicle]["total_speed"]+=t['speed']
        summary[vehicle]["violations_count"]+=len(t['violations'])
    print("Vehicle Category Summary:")
    for name,data  in summary.items():
        avg_speed=data["total_speed"]/data["count"]
        print(f"{name}: Count={data['count']}, Avg Speed={avg_speed:}, Violations={data['violations_count']}")
vehicle_summary(structured_data)
```

```python

print("Task 8: ")
def high_congestion(traffic):
    high={}
    for t in traffic:
        z=t['zone']
        if z not in high:
            high[z]=0
        if t['status']=="Congested":
            high[z]+=1
    hzone=max(high,key=high.get)
    hcount=high[hzone]
    samecount=[]
    for k,v in high.items():
        if v==hcount:
            samecount.append(k)
    print("Zones with congestion: ",samecount)
high_congestion(structured_data)
```

```python
print("Task 9: ")
def timewindow(traffic):
    for t in traffic:
        hour = int(t['time'].split(':')[0]) 
        if hour>= 6 and hour<= 11:
            window = "Morning"
        elif hour>= 12 and hour<= 16:
            window = "Afternoon"
        elif hour>= 17 and hour<= 20:
            window = "Evening"
        else:
            window = "Night"
        t['time_window']=window
    for t in traffic:
        print(f" VehicleID {t['id']} Type:{t['vehicle']} Time: {t['time']} Window: {t['time_window']}")
timewindow(structured_data)
```

```python
print("Task 10: ")
def zone_report(traffic):
    zone_summary = {}
    for t in traffic:
        z = t['zone']
        if z not in zone_summary:
            zone_summary[z] = {"vehicles": 0,"total_speed": 0,"violations_count": 0,"vehicle_types": {}}
        zone_summary[z]["vehicles"] += 1
        zone_summary[z]["total_speed"] += t['speed']
        zone_summary[z]["violations_count"] += len(t['violations'])
        vt = t['vehicle']
        if vt not in zone_summary[z]["vehicle_types"]:
            zone_summary[z]["vehicle_types"][vt] = 0
        zone_summary[z]["vehicle_types"][vt] += 1

    print("Final Zone-Level Report:")
    for z, data in zone_summary.items():
        avg_speed =data["total_speed"]/data["vehicles"]
        common_vehicle = max(data["vehicle_types"], key=data["vehicle_types"].get)
        if data["violations_count"] == 0:
            safety_cat = "Excellent"
        elif data["violations_count"] <= 2:
            safety_cat = "Good"
        else:
            safety_cat = "Needs Improvement"
        print(f"Zone {z}:")
        print(f"  Total Vehicles: {data['vehicles']}")
        print(f"  Average Speed: {avg_speed:}")
        print(f"  Total Violations: {data['violations_count']}")
        print(f"  Most Common Vehicle: {common_vehicle}")
        print(f"  Safety Category: {safety_cat}\n")
zone_report(structured_data)
```
