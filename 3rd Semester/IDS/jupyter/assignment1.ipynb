```python
#5. problem: write a function
"""
Given a year, determine whether it is a leap year. 
If it is a leap year, return the Boolean True, otherwise return False.

Note that the code stub provided reads from STDIN 
and passes arguments to the is_leap function. 
It is only necessary to complete the is_leap function.
"""
def is_leap(year):
    leap = False
    if year%400==0:
        leap=True
    elif year%100==0:
        leap=False
    elif year%4==0:
        leap=True
        
    return leap

year = int(input())
print(is_leap(year))
```

For this task, I was required to determine whether a given year is a leap year. To solve this, I implemented a simple conditional check following the Gregorian calendar rules. I first checked if the year is divisible by 4. If it was, I then verified whether it was also divisible by 100. In that case, the year would not be a leap year unless it was further divisible by 400, in which case it would be. Using these logical conditions, the program correctly identifies and returns True for leap years and False otherwise.

```python
#8. problem: find the runner up score
"""
Given the participants' score sheet for your University Sports Day,
 you are required to find the runner-up score. You are given  scores. 
 Store them in a list and find the score of the runner-up.
"""
if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    
    unique_arr=set(arr)
    
    sorted_unique=sorted(unique_arr)
    
    runner_up=sorted_unique[-2]
    
    print(runner_up)
```

For this task, I needed to find the runner-up score from a list of participant scores. My approach was to first read all the scores and remove duplicates using a set() to ensure only unique scores remained. I then sorted these scores in ascending order and selected the second highest score using index [-2]. This value represented the runner-up score, which was then printed as the output. The logic ensures that even if multiple participants share the same top score, the correct second highest unique score is identified.

```python
#9. problem: nested lists
"""
Given the names and grades for each student in a class of  students, 
store them in a nested list and print the name(s) of any student(s) having 
the second lowest grade.
Note: If there are multiple students with the second lowest grade, 
order their names alphabetically and print each name on a new line.
"""
if __name__ == '__main__':
    records = []
    for _ in range(int(input())):
        name = input()
        score = float(input())
        records.append([name, score])
    scores = [score for name, score in records]
    unq_scr=set(scores)
    unq_scr=sorted(unq_scr)
    second_lowest = unq_scr[1]
    names = [name for name, score in records 
                    if score == second_lowest]
    #names=list of all names with same second lowest score
    for name in sorted(names):
        print(name)
```

In this task, I was required to identify the students who had the second lowest grade. I began by storing each student’s name and grade as pairs in a nested list. Then, I extracted all the grades, removed duplicates using a set(), and sorted them to find the second lowest unique grade. After identifying that grade, I collected the names of all students who achieved it. Finally, I sorted these names alphabetically and printed each one on a new line. This logic ensured accurate identification of all students sharing the second lowest grade

```python
#10. problem: finding the percentage
"""
The provided code stub will read in a dictionary containing key/value 
pairs of name:[marks] for a list of students. Print the average of the 
marks array for the student name provided, showing 2 places after the decimal
"""
if __name__ == '__main__':
    n = int(input())  
    student_marks = {}
    
    for _ in range(n):
        # split input line into name and marks
        line = input().split()
        name = line[0]
        scores = list(map(float, line[1:]))  
        student_marks[name] = scores  # store in dictionary

    query_name = input()  # name to query
    avg = sum(student_marks[query_name]) / len(student_marks[query_name])
    
    print(f"{avg:.2f}")  # print average with 2 decimal places
```

For this task, I was asked to calculate and display the average marks of a specific student. I stored each student’s name and their list of marks in a dictionary, where the student’s name served as the key and their marks as the values. Once the query name was entered, I retrieved the corresponding list of marks and calculated the average by dividing the sum of the marks by the number of subjects. Lastly, I displayed the average rounded to two decimal places using formatted output. This approach ensured accurate computation and clear presentation of the student’s average marks
