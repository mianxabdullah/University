```python
students_data = [
 "id:101,name:Ali,dept:CS,attendance:88,marks:[80,75,90],feedback:Excellent",
 "id:102,name:Sara,dept:IT,attendance:72,marks:[60,65,58],feedback:Helpful",
 "id:103,name:Hamza,dept:CS,attendance:95,marks:[90,92,88],feedback:Very Good",
 "id:104,name:Maryam,dept:Math,attendance:67,marks:[55,60,50],feedback:Average",
 "id:105,name:Usman,dept:IT,attendance:85,marks:[70,68,72],feedback:Helpful"
]
#task1
stud=[]
for s in students_data:
    try:
        parts=s.split(',',4)
        if "marks" in parts[-1]:
            m,f = parts[-1].split("],")
            parts[-1] = m + "]"      
            parts.append(f)         
        d={}
        for p in parts:
            key,val=p.split(':',1)
            if key in ["id","attendance"]:
                val=int(val)
            elif key == "marks":
                val = val.strip("[]")
                val = [int(x) for x in val.split(',')]
            d[key]=val
        stud.append(d)
    except Exception as e:
        print("Error in record:", s, "->", e)
print(stud)
#task2
def avgAttendByDept(students):
    cs=[]
    it=[]
    math=[]
    for s in students:
        if s['dept']=="CS":
            cs.append(s['attendance'])
        elif s['dept']=="IT":
            it.append(s['attendance'])
        elif s['dept']=="Math":
            math.append(s['attendance'])
    cs_avg=sum(cs)/len(cs)
    it_avg=sum(it)/len(it)
    math_avg=sum(math)/len(math)
    print("Dept Wise Avg Attendance: ")
    print("CS: ",cs_avg)
    print("IT: ",it_avg)
    print("Math: ",math_avg)

avgAttendByDept(stud)
#task3
def low_attendance(students):
    low=[]   
    for s in students:
        if s['attendance']<75:
            low.append(s['name'])
    return low
print(low_attendance(stud))
#task4
def avgMarks(students):
    avg={}
    for s in students:
        marks_sum=sum(s['marks'])
        avg[s['name']]=marks_sum/len(s['marks'])
    return avg
print(avgMarks(stud))
#task5
listcs=[]
listit=[]
listmath=[]
for s in stud:
    if s['dept']=="CS":
        listcs.append(s)
    elif s['dept']=="IT":
        listit.append(s)
    elif s['dept']=="Math":
        listmath.append(s)
csavg=avgMarks(listcs)
itavg=avgMarks(listit)
mathavg=avgMarks(listmath)
highcs=max(csavg,key=csavg.get) #max(csavg,key=csavg.get) returns name of student with highest in csavg dict
highit=max(itavg,key=itavg.get)
highmath=max(mathavg,key=mathavg.get)
print("Top Students by Department:")
print("CS:", highcs, "with average marks", csavg[highcs])
print("IT:", highit, "with average marks", itavg[highit])
print("Math:", highmath, "with average marks", mathavg[highmath])
"""
#task6 we cant perform this task because we dont have courses in data set
all_courses = []
for s in stud:
    # assuming each student has s['courses'] = ['CS101', 'Math101', ...]
    all_courses.extend(s['courses'])
unique_courses = set(all_courses)  # removes duplicates
count = len(unique_courses)
print("Total unique courses offered:", count)"""
#task7
feedbacks=[]
count=0
for s in stud:
    comment=s['feedback'].lower()
    feedbacks.append(comment)
    if "helpful" in comment:
        count+=1
print("count of comments: ",count)
print("list of comments: ",feedbacks)
#task8
averages=avgMarks(stud)
category={}
for key,val in averages.items():
    if val>=85:
        category[key]="excellent"
    elif val>=70 and val<85:
        category[key]="good"
    elif val>=60 and val<70:
        category[key]="average"
    elif val<60:
        category[key]="needs improvement"
print(category)
#task9
def avgAttendance(depart):
    dpt=[]
    for d in depart:
        dpt.append(d['attendance'])
    avg=sum(dpt)/len(dpt)
    return avg

def summary(dept):
    report={}
    report['name']=dept[0]['dept']
    report['no_of_studs']=len(dept)
    report['attendance']=avgAttendance(dept)
    avgmark=avgMarks(dept)
    topstud=max(avgmark,key=avgmark.get)
    report['top_student']=topstud
    return report
print("Department wise Summary: ")
print(summary(listcs))
print(summary(listit))
print(summary(listmath))
#task10
""" we cant perform the task upon given data set
courses = ['CS101', 'Math101', 'IT101']
students = [s['name'] for s in stud]
num_pairs = min(len(students), len(courses))
pairs = {}
for i in range(num_pairs):
    pairs[students[i]] = courses[i]
print("Student-Course Pairs:")
print(pairs)
#handle leftover studs or courses
if len(students) > len(courses):
    print("\nWarning: Not enough courses for all students. Unassigned students:", students[num_pairs:])
elif len(courses) > len(students):
    print("\nWarning: More courses than students. Unassigned courses:", courses[num_pairs:])

"""
```
