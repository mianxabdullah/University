```python
  # write aprogram that looks for the string "leave" in each line of he file "test.txt" and prints found if string is found in that lineor not found if string is not found in that line 
fhand = open('test.txt')
for line in fhand:
    line = line.rstrip()
    if 'leave' in line:
      print('found')
    else:
      print('not found')
```
