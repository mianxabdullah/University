```python
# find the purpose of math.copysign(x,y) function and run it with sample data

x = 5.0
y = -3.0
result = math.copysign(x, y)
print(f"Result of copysign({x}, {y}): {result}")
# purpose of math.copysign(x,y) function is to return a float with the magnitude (absolute value) of x but the sign of y. In this case, it returns -5.0 because x is 5.0 and y is negative.
```

```python
# find the purpose of math.trunc(x) function and run it with sample data

x = 5.7
result = math.trunc(x)
print(f"Result of trunc({x}): {result}")
# purpose of math.trunc(x) function is to return the integer part of x, effectively truncating the decimal portion. In this case, it returns 5 because it removes the decimal part of 5.7.
```

```python
# find the purpose of math.fabs(x) function and run it with sample data

x = -7.5
result = math.fabs(x)
print(f"Result of fabs({x}): {result}")
# purpose of math.fabs(x) function is to return the absolute value of x as a float. In this case, it returns 7.5 because it removes the negative sign from -7.5.
```

# 4. Adding new functions

```python
def print_lyrics():
    print ("I'm a lumberjack, and I'm okay.")
    print ('I sleep all night and I work all day.')
```

```python
print (print_lyrics)
type(print_lyrics)
```

```python
print_lyrics()
```

```python
def repeat_lyrics():
    print_lyrics()
    print_lyrics()
```

```python
repeat_lyrics()
```

```python
# parameterise function
def print_twice(bruce):
    print (bruce)
    print (bruce)
```

```python
print_twice('hi')
```

```python
print_twice(3)
```

```python
print_twice("pak" * 3) # The argument is evaluated first, before the function is called
```

```python
# To return a result from a function, we use the return statement in our function.
def add(a, b):
    added = a + b
    return added
```

```python
print (add(3,4))
```

```python
# define a function, to which when your name is passed as an argument, returns the no of characters in your name
def count_characters(name):
    return len(name)    

# Test the function
name = "Alice"
print(f"Number of characters in {name}: {count_characters(name)}")
```

```python
# define a function to find the maximum of three numbers
def find_maximum(a, b, c):
    return max(a, b, c)

# Test the function
a = 10
b = 20
c = 30
print(f"Maximum of {a}, {b}, and {c}: {find_maximum(a, b, c)}")
```

```python
# define a function that accepts an argument which is a string and you have to reverse that string
def reverse_string(s):
    return s[::-1]

# Test the function
s = "Hello, World!"
print(f"Reversed string of '{s}': '{reverse_string(s)}'")
```
